```python
from nltk.corpus import wordnet
import nltk

# nltk.download('wordnet')

# Get all lemmas from WordNet
lemmas = set()
for synset in wordnet.all_synsets():
    for lemma in synset.lemmas():
        lemmas.add(lemma.name())

# Filter for consonant + s pattern
pattern = r'[a-zA-Z]*[ebdfghjklmnpqrtvwxz]s\b'
suspect_lemmas = [lemma for lemma in lemmas if re.match(pattern, lemma)]
suspect_lemmas[:10]
```

```python
def clean_s(word):
    pattern = r'(\b[a-zA-Z]*[bcdfghjklmnpqrtvwxz])(s)\b'
    if re.match(pattern, word):
        if word[-3:] not in ["ers", "ens", "ins", "els", "lds", "bbs", "wts", "ors"]:
            word = re.sub(pattern, r'\1', word)
    return word
```

```python
clean_s("Dickens")
```

```python
mostfrequent10000 = [item[1]["lemmata"].lower() for item in unigrams_lemmata_dict.items()]
mostfrequent10000[:10]
```

```python
for item in unigrams_lemmata_dict.items():
    val_dict = item[1]
    word = val_dict["lemmata"].lower()
    word = clean_s(word)
    if word in mostfrequent10000:
        val_dict["lemmata"] = word
        unigrams_lemmata_dict[item[0]] = val_dict
```
