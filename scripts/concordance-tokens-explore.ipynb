```python
_token_re = re.compile(r"\w+", flags=re.UNICODE)

hits["kwic_tokens"] = [
    _token_re.findall(f"{l} {r}")
    for l, r in zip(hits["kwic_left"].fillna(""), hits["kwic_right"].fillna(""))
]
```
